# Breaking apart multi valued attributes into tables and setting appropriate metadata

```python
#business to business_category table
with engine.begin() as conn:
    conn.exec_driver_sql("""
    IF OBJECT_ID('business_category','U') IS NOT NULL DROP TABLE business_category;
    CREATE TABLE business_category (
        business_id VARCHAR(50),
        category NVARCHAR(200),
        PRIMARY KEY (business_id, category),
        FOREIGN KEY (business_id) REFERENCES business(business_id)
    );
    """)

business_category = (
    business_df
    .assign(category=business_df["categories"].str.split(","))
    .explode("category")
)

business_category["category"] = business_category["category"].str.strip()
business_category = business_category[["business_id", "category"]].dropna().drop_duplicates()

business_category.to_sql(
    "business_category",
    engine,
    if_exists="append",
    index=False,
    method="multi",
    chunksize=500
)

print("✅ business_category completed")

```

```python
#business to business_attributes table

with engine.begin() as conn:
    conn.exec_driver_sql("""
    IF OBJECT_ID('business_attributes','U') IS NOT NULL DROP TABLE business_attributes;
    CREATE TABLE business_attributes (
        business_id VARCHAR(50),
        attr_key NVARCHAR(200),
        attr_value NVARCHAR(MAX),
        PRIMARY KEY (business_id, attr_key),
        FOREIGN KEY (business_id) REFERENCES business(business_id)
    );
    """)

import json

rows = []
for idx, row in business_df.iterrows():
    if isinstance(row["attributes"], dict):
        for k, v in row["attributes"].items():
            rows.append((row["business_id"], k, json.dumps(v)))

business_attributes = pd.DataFrame(rows, columns=["business_id","attr_key","attr_value"])

business_attributes.to_sql(
    "business_attributes",
    engine,
    if_exists="append",
    index=False,
    method="multi",
    chunksize=500
)

print("✅ business_attributes completed")
```

```python
#business to business_hours

with engine.begin() as conn:
    conn.exec_driver_sql("""
    IF OBJECT_ID('business_hours','U') IS NOT NULL DROP TABLE business_hours;
    CREATE TABLE business_hours (
        business_id VARCHAR(50),
        day_name NVARCHAR(50),
        hours_range NVARCHAR(200),
        PRIMARY KEY (business_id, day_name),
        FOREIGN KEY (business_id) REFERENCES business(business_id)
    );
    """)

rows = []
for idx, row in business_df.iterrows():
    if isinstance(row["hours"], dict):
        for day, hours in row["hours"].items():
            rows.append((row["business_id"], day, hours))

business_hours = pd.DataFrame(rows, columns=["business_id","day_name","hours_range"])

business_hours.to_sql(
    "business_hours",
    engine,
    if_exists="append",
    index=False,
    method="multi",
    chunksize=500
)

print("✅ business_hours completed")
```

```python
#users to users_elite

with engine.begin() as conn:
    conn.exec_driver_sql("""
    IF OBJECT_ID('user_elite','U') IS NOT NULL DROP TABLE user_elite;
    CREATE TABLE user_elite (
        user_id VARCHAR(50),
        elite_year INT,
        PRIMARY KEY (user_id, elite_year),
        FOREIGN KEY (user_id) REFERENCES user_account(user_id)
    );
    """)

user_elite = user_df.copy()
user_elite["elite"] = user_elite["elite"].astype(str).str.split(",")

user_elite = user_elite.explode("elite")
user_elite = user_elite[user_elite["elite"].str.strip() != ""]
user_elite["elite"] = pd.to_numeric(user_elite["elite"], errors="coerce")
user_elite = user_elite.dropna(subset=["elite"])[["user_id","elite"]].drop_duplicates()

user_elite.columns = ["user_id","elite_year"]

user_elite.to_sql(
    "user_elite",
    engine,
    if_exists="append",
    index=False,
    method="multi",
    chunksize=500
)

print("✅ user_elite completed")
```

```python
#user to user_friend

with engine.begin() as conn:
    conn.exec_driver_sql("""
    IF OBJECT_ID('user_friend','U') IS NOT NULL DROP TABLE user_friend;
    CREATE TABLE user_friend (
        user_id VARCHAR(50),
        friend_user_id VARCHAR(50),
        PRIMARY KEY (user_id, friend_user_id),
        FOREIGN KEY (user_id) REFERENCES user_account(user_id)
    );
    """)

user_friend = user_df.copy()
user_friend['friends'] = user_friend['friends'].astype(str).str.split(',')

user_friend = user_friend.explode('friends')
user_friend = user_friend[user_friend['friends'].str.strip() != ""]

user_friend = user_friend[['user_id','friends']].drop_duplicates()
user_friend.columns = ["user_id","friend_user_id"]

user_friend = user_friend.head(100000)

user_friend.to_sql(
    "user_friend",
    engine,
    if_exists="append",
    index=False,
    method="multi",
    chunksize=500
)

print("✅ user_friend completed")
```

**6. Most Common Positive Words in Reviews** (Purpose: Identify what customers appreciate most about Chinese restaurants.)

SELECT LOWER(word) AS keyword, COUNT(*) AS frequencyFROM (
  SELECT TRIM(REPLACE(REPLACE(REPLACE(text, '.', ''), ',', ''), '!', '')) AS text_clean
  FROM review
  JOIN business USING(business_id)
  WHERE city='Chicago' AND categories LIKE '%Chinese%' AND stars >= 4)CROSS JOIN (
  SELECT value AS word FROM json_each('["good","fresh","tasty","service","fast","friendly","healthy","clean","portion","flavor"]')
)WHERE text_clean LIKE '%' || word || '%' GROUP BY keywordORDER BY frequency DESC;

**7. Most Common Complaint Words (Negative Reviews)**(Purpose: Understand common pain points you can improve on.)

SELECT TOP 10 LOWER(keyword) AS keyword, COUNT(*) AS frequencyFROM (
    SELECT CASEWHEN text LIKE '%slow%' THEN 'slow'WHEN text LIKE '%cold%' THEN 'cold'WHEN text LIKE '%expensive%' THEN 'expensive'WHEN text LIKE '%rude%' THEN 'rude'WHEN text LIKE '%greasy%' THEN 'greasy'ELSE NULL END AS keyword
    FROM review r
    JOIN business b ON r.business_id = b.business_id
    WHERE b.city = 'Chicago'AND b.categories LIKE '%Chinese%'AND r.stars <= 2) AS keywordsWHERE keyword IS NOT NULL GROUP BY keywordORDER BY frequency DESC;

**8. Average Review Sentiment per Rating**(Purpose: Long positive reviews can signal strong emotional connection or loyalty.)

SELECT stars, ROUND(AVG(LENGTH(text)), 2) AS avg_review_length, COUNT(*) AS num_reviewsFROM reviewJOIN business USING(business_id)WHERE city='Chicago' AND categories LIKE '%Chinese%' GROUP BY starsORDER BY stars DESC;

SELECT r.review_id, b.name AS business_name, r.text, r.useful, r.starsFROM review rJOIN business b ON r.business_id = b.business_idWHERE b.city='Chicago' AND b.categories LIKE '%Chinese%' ORDER BY r.useful DESCLIMIT 10;
